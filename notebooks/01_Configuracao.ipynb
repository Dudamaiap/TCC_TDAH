# 01 Configuração
Este notebook prepara o ambiente do projeto TDAH, cria a estrutura de pastas, verifica as dependÃªncias utilizadas nos notebooks posteriores e fixa as sementes de aleatoriedade para favorecer a reprodutibilidade.

Antes de executar

No terminal do VS Code, na pasta raiz do projeto:

python -m venv .venv

.venv\Scripts\activate       # Windows

source .venv/bin/activate  # Linux/Mac

pip install -r requirements.txt

Depois, selecione o kernel .venv no canto superior direito do VS Code.

```python
from pathlib import Path
import sys
import importlib

# Localizar a raiz do projeto procurando o config.py
RAIZ = Path.cwd()
while not (RAIZ / "config.py").exists() and RAIZ != RAIZ.parent:
    RAIZ = RAIZ.parent

if not (RAIZ / "config.py").exists():
    raise FileNotFoundError(
        "config.py não foi encontrado. Verifique se ele está na raiz do projeto TDAH."
    )

# Permitir importação do config.py mesmo com o notebook dentro de notebooks/
if str(RAIZ) not in sys.path:
    sys.path.insert(0, str(RAIZ))

import config
importlib.reload(config)

# Criar estrutura de pastas
config.criar_pastas()


print("CONFIGURAÇÃO DO PROJETO TDAH")

print("Raiz do projeto:", config.RAIZ)
print("Python:", sys.version.split()[0])
print("Seed:", config.SEED)
print("Número de ROIs:", config.N_ROIS)
print("Número de conexões:", config.N_CONNECTIONS)
print("Número de folds:", config.N_FOLDS)
```

```python
# Verificar dependências

pacotes = [
    "numpy", "pandas", "scipy", "sklearn", "nilearn",
    "matplotlib", "seaborn", "tqdm", "torch", "joblib", "xgboost"
]

faltando = []

for pacote in pacotes:
    try:
        modulo = importlib.import_module(pacote)
        versao = getattr(modulo, "__version__", "versão não informada")
        print(f" {pacote}: {versao}")
    except ImportError:
        print(f" {pacote}: não instalado")
        faltando.append(pacote)

print("\nPacotes faltando:", faltando or "nenhum")

if faltando:
    print("\nInstale as dependências com:")
    print("python -m pip install -r requirements.txt")
```

```python
# Configurar sementes para reprodutibilidade

import random
import numpy as np

random.seed(config.SEED)
np.random.seed(config.SEED)

print("Seed Python:", config.SEED)
print("Seed NumPy:", config.SEED)

try:
    import torch

    torch.manual_seed(config.SEED)

    if torch.cuda.is_available():
        torch.cuda.manual_seed(config.SEED)
        torch.cuda.manual_seed_all(config.SEED)

    print("PyTorch:", torch.__version__)
    print("CUDA disponível:", torch.cuda.is_available())

    if torch.cuda.is_available():
        print("GPU:", torch.cuda.get_device_name(0))
    else:
        print("GPU CUDA não disponível. PyTorch utilizará CPU.")

except ImportError:
    print("PyTorch não instalado.")
```

```python
# Verificar a estrutura criada

print("ESTRUTURA DO PROJETO")


pastas = {
    "data": config.DATA,
    "data/raw": config.RAW,
    "data/extracted": config.EXTRACTED,
    "data/processed": config.PROCESSED,
    "data/connectivity": config.CONNECTIVITY,
    "docs": config.DOCS,
    "figures": config.FIGURES,
    "models": config.MODELS,
    "notebooks": config.NOTEBOOKS,
    "results": config.RESULTS,
}

for nome, caminho in pastas.items():
    status = "OK" if caminho.exists() else "ERRO"
    print(f"[{status}] {nome}: {caminho}")

print("CONFIGURAÇÃO CONCLUÍDA")
```
